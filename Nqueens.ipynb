```python
class solution:

    def __init__(self):
        self.x = []
        self.found = False

    def place(self, k, i):
        for j in range(1, k):
            if self.x[j] == i or abs(self.x[j] - i) == abs(j - k):
                return False
        return True

    def printplacedqueen(self, N):
        print("Solution:")
        for i in range(1, N + 1):
            for j in range(1, N + 1):
                if self.x[i] == j:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()

    def N_Queens(self, i, N):
        if i == N + 1:
            self.printplacedqueen(N)
            self.found = True
            return

        for j in range(1, N + 1):
            if self.place(i, j):
                self.x[i] = j
                self.N_Queens(i + 1, N)

                if self.found:
                    return


N = int(input("Enter the queens value: "))

obj = solution()
obj.x = [0] * (N + 1)

obj.N_Queens(1, N)

if not obj.found:
    print("No solution exists")
```
